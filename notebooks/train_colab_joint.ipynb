## 5. 모델 초기화

```python
import torch
import torch.nn as nn
from pathlib import Path

from model.model import KnobNetJoint
from utils.config import KNOB_PARAMS

device = torch.device("cuda" if torch.cuda.is_available() else "cpu")
print("device:", device)

model = KnobNetJoint(num_knobs=len(KNOB_PARAMS), freeze_mert=True).to(device)
model.summary()
```

## 6. Phase 1: Head만 학습 (MERT frozen)

```python
from train.cache import (
    run_epoch_cached, evaluate_all_cached,
)
from train.train import (
    make_optimizer, save_checkpoint, load_checkpoint,
    log_epoch, log_param_mae, log_accuracy, log_csv,
)

CKPT_DIR = Path("/content/drive/MyDrive/KnobNet/checkpoints_joint")
CKPT_DIR.mkdir(parents=True, exist_ok=True)
CKPT_P1 = CKPT_DIR / "phase1_best.pt"
LOG_CSV  = CKPT_DIR / "training_log.csv"

PHASE1_EPOCHS = 60
criterion = nn.L1Loss()
optimizer = make_optimizer(model, lr=1e-3, phase=1)
scheduler = torch.optim.lr_scheduler.CosineAnnealingLR(optimizer, T_max=PHASE1_EPOCHS)
scaler    = torch.amp.GradScaler("cuda") if device.type == "cuda" else None

start_epoch = 1
best_val = float("inf")
if CKPT_P1.exists():
    meta = load_checkpoint(CKPT_P1, model, optimizer, scheduler)
    start_epoch = meta["epoch"] + 1
    best_val = meta["val_loss"]
    print(f"체크포인트 재개: epoch={meta['epoch']}  best_val={best_val:.4f}")

print(f"Phase 1 학습: epoch {start_epoch} ~ {PHASE1_EPOCHS}  (AMP={'on' if scaler else 'off'})")
model.summary()
```

```python
if start_epoch == 1 and LOG_CSV.exists():
    LOG_CSV.unlink()
    print("기존 CSV 초기화 (새 학습 시작)")

for epoch in range(start_epoch, PHASE1_EPOCHS + 1):
    train_loss = run_epoch_cached(model, train_loader_c, optimizer, criterion, device, scaler)
    metrics    = evaluate_all_cached(model, val_loader_c, criterion, device, tolerance=0.1)
    val_loss   = metrics["val_loss"]
    scheduler.step()

    improved = val_loss < best_val
    if improved:
        best_val = val_loss
        save_checkpoint(CKPT_P1, model, epoch, phase=1, val_loss=val_loss,
                        optimizer=optimizer, scheduler=scheduler)

    log_epoch(1, epoch, PHASE1_EPOCHS, train_loss, val_loss, improved)
    log_param_mae(metrics["mae"])
    log_accuracy(metrics["acc"], tolerance=0.1)
    log_csv(LOG_CSV, 1, epoch, train_loss, val_loss, metrics["mae"], metrics["acc"])

print("\nPhase 1 완료")
```

## 7. Phase 2: MERT fine-tuning

```python
from train.train import run_epoch, evaluate_all

CKPT_P2 = CKPT_DIR / "phase2_best.pt"
PHASE2_EPOCHS = 12

load_checkpoint(CKPT_P1, model)

model.unfreeze_mert()
optimizer = make_optimizer(model, lr=1e-3, phase=2)
scheduler = torch.optim.lr_scheduler.CosineAnnealingLR(optimizer, T_max=PHASE2_EPOCHS)
scaler    = torch.amp.GradScaler("cuda") if device.type == "cuda" else None

start_epoch = 1
best_val = float("inf")
if CKPT_P2.exists():
    meta = load_checkpoint(CKPT_P2, model, optimizer, scheduler)
    start_epoch = meta["epoch"] + 1
    best_val = meta["val_loss"]
    print(f"Phase 2 재개: epoch={meta['epoch']}  best_val={best_val:.4f}")

print(f"Phase 2 학습: epoch {start_epoch} ~ {PHASE2_EPOCHS}  (AMP={'on' if scaler else 'off'})")
model.summary()
```

```python
for epoch in range(start_epoch, PHASE2_EPOCHS + 1):
    train_loss = run_epoch(model, train_loader, optimizer, criterion, device, scaler)
    metrics    = evaluate_all(model, val_loader, criterion, device, tolerance=0.1)
    val_loss   = metrics["val_loss"]
    scheduler.step()

    improved = val_loss < best_val
    if improved:
        best_val = val_loss
        save_checkpoint(CKPT_P2, model, epoch, phase=2, val_loss=val_loss,
                        optimizer=optimizer, scheduler=scheduler)

    log_epoch(2, epoch, PHASE2_EPOCHS, train_loss, val_loss, improved)
    log_param_mae(metrics["mae"])
    log_accuracy(metrics["acc"], tolerance=0.1)
    log_csv(LOG_CSV, 2, epoch, train_loss, val_loss, metrics["mae"], metrics["acc"])

print("\nPhase 2 완료")
```
